```python
marks=float(input("enter your marks:"))

if marks>=90:
  grade="A"
elif marks>=75 and marks<=89:
  grade="B"
elif marks>=50 and marks<=74:
  grade="C"

else:
  grade="F"
  print("Failed")

print("grade of the student:",grade)
```
