```python
#pros and cons of One-Hot Encoding and Label Encoding.

#One-Hot Encoding
#Pros

#No ordinal relationship is assumed between categories

#Works very well with linear and distance-based models

#Prevents misleading numeric ordering

#Cons

#Increases dimensionality (many new columns)

#Not suitable for high-cardinality features

#Can increase memory usage and training time
```

```python
#Label Encoding
#Pros

#Simple and memory efficient

#Suitable for high-cardinality categorical features

#Works well with tree-based models

#Cons

#Introduces artificial order between categories

#Can mislead linear and distance-based algorithms

#Numeric labels may be misinterpreted as magnitude
```

```python
#Feature 1: Age Group
def age_group(age):
    if age <= 30:
        return "Young"
    elif age <= 50:
        return "Middle_Aged"
    else:
        return "Senior"

df["age_group"] = df["age"].apply(age_group)

# Feature 2: Total Capital
df["total_capital"] = df["capital_gain"] - df["capital_loss"]

# Show output
df[["age", "age_group", "capital_gain", "capital_loss", "total_capital"]].head()
```
