```python
#W=1000+5*X
# X=mu=600
# W=1000+5(600)
# W=mu=4000

# H0: mu=4000
# H1: mu!=4000
```

```python
from scipy.stats import norm
x_bar=3050
μ =4000
σ = 5*25
n=25
alpha=0.05
z=x_bar-μ
se=σ/n**0.5
z=(x_bar-μ)/se
z_critical=norm.ppf(1-alpha)
print('test statistic z value:',z)
print('critical z:',z_critical)
if z>z_critical:
    print('It accepts Alternate Hypothesis')
else:
    print('It accepts Null Hypothesis ')
```
